```python
scores_if = -isolation_forest.decision_function(X)

print("Minimum score:", scores_if.min())
print("Maximum score:", scores_if.max())
```

```python
thresholds = np.percentile(scores_if, np.arange(90, 100, 0.5))

results = []

for threshold in thresholds:
    predictions = (scores_if >= threshold).astype(int)

    precision = precision_score(y, predictions, zero_division=0)
    recall = recall_score(y, predictions, zero_division=0)
    f1 = f1_score(y, predictions, zero_division=0)

    results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

threshold_df = pd.DataFrame(results)

threshold_df.head()
```

```python
best_result = threshold_df.loc[
    threshold_df["f1"].idxmax()
]

print("Best Threshold:", best_result["threshold"])
print("Precision:", best_result["precision"])
print("Recall:", best_result["recall"])
print("F1 Score:", best_result["f1"])
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score

# Isolation Forest anomaly score
# Higher score = more anomalous
scores_if = -isolation_forest.decision_function(X)

# Different thresholds test karenge
thresholds = np.percentile(
    scores_if,
    np.arange(90, 100, 0.5)
)

results = []

for threshold in thresholds:

    predictions = (scores_if >= threshold).astype(int)

    precision = precision_score(
        y,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y,
        predictions,
        zero_division=0
    )

    results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

threshold_df = pd.DataFrame(results)

print("Threshold tuning completed! ✅")
print("Number of thresholds tested:", len(threshold_df))
```

```python
print(
    threshold_df
    .sort_values("f1", ascending=False)
    .head(10)
)
```

```python
best_threshold = threshold_df.loc[
    threshold_df["f1"].idxmax(),
    "threshold"
]

final_predictions = (scores_if >= best_threshold).astype(int)

final_precision = precision_score(
    y,
    final_predictions,
    zero_division=0
)

final_recall = recall_score(
    y,
    final_predictions,
    zero_division=0
)

final_f1 = f1_score(
    y,
    final_predictions,
    zero_division=0
)

print("FINAL ISOLATION FOREST RESULTS")
print("--------------------------------")
print(f"Best Threshold: {best_threshold:.6f}")
print(f"Precision:      {final_precision:.4f}")
print(f"Recall:         {final_recall:.4f}")
print(f"F1 Score:       {final_f1:.4f}")
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score
import matplotlib.pyplot as plt

# Precision-Recall curve
precision_curve, recall_curve, pr_thresholds = precision_recall_curve(
    y,
    scores_if
)

# Average Precision Score
average_precision = average_precision_score(y, scores_if)

# Plot PR Curve
plt.figure(figsize=(8, 6))

plt.plot(
    recall_curve,
    precision_curve,
    label=f"Isolation Forest (AP = {average_precision:.4f})"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve - Isolation Forest")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Find the PR point closest to our best tested threshold

best_idx = np.argmin(
    np.abs(pr_thresholds - best_threshold)
)

best_precision_pr = precision_curve[best_idx]
best_recall_pr = recall_curve[best_idx]

plt.figure(figsize=(8, 6))

plt.plot(
    recall_curve,
    precision_curve,
    label="Isolation Forest PR Curve"
)

plt.scatter(
    best_recall_pr,
    best_precision_pr,
    s=100,
    label=f"Best Tested Threshold = {best_threshold:.6f}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve with Best Tested Threshold")
plt.legend()
plt.grid(True)

plt.show()

print("BEST TESTED THRESHOLD")
print("----------------------")
print(f"Threshold : {best_threshold:.6f}")
print(f"Precision : {final_precision:.4f}")
print(f"Recall    : {final_recall:.4f}")
print(f"F1 Score  : {final_f1:.4f}")
```
